# EduPro Course Demand & Revenue Forecasting - Business Insights
This notebook summarizes key findings, pricing suggestions, new category expansion plans, and teacher hiring insights based on our EDA and models.

```python
import pandas as pd
import numpy as np
from IPython.display import display

# Load datasets
course_level = pd.read_csv('../data/processed/course_level_dataset.csv')
```

### 1. Pricing Strategy Insights

```python
pricing_summary = course_level.groupby('PriceBand', observed=False).agg(
    CourseCount=('CourseID', 'count'),
    AvgEnrollments=('TotalEnrollments', 'mean'),
    AvgRevenue=('TotalRevenue', 'mean')
).reset_index()
display(pricing_summary)
```

### 2. Category Performance Insights

```python
category_summary = course_level.groupby('CourseCategory').agg(
    CourseCount=('CourseID', 'count'),
    AvgEnrollments=('TotalEnrollments', 'mean'),
    TotalRevenue=('TotalRevenue', 'sum')
).sort_values(by='TotalRevenue', ascending=False)
display(category_summary)
```

### 3. Instructor Insights

```python
experience_summary = course_level.groupby('ExperienceBucket', observed=False).agg(
    AvgRating=('TeacherRating', 'mean'),
    AvgEnrollments=('TotalEnrollments', 'mean'),
    TotalRevenue=('TotalRevenue', 'sum')
)
display(experience_summary)
```
