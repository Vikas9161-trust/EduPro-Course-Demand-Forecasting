# EduPro Course Demand & Revenue Forecasting - Data Understanding
This notebook loads the raw Excel sheet and inspects its sheets, shapes, and columns.

```python
import pandas as pd
excel_path = r"../data/raw/EduPro_Dataset.xlsx"
xl = pd.ExcelFile(excel_path)
print("Sheets in spreadsheet:", xl.sheet_names)
```

### Inspecting individual sheets

```python
for sheet_name in xl.sheet_names:
    df = xl.parse(sheet_name)
    print(f"\n--- {sheet_name} ---")
    print(f"Shape: {df.shape}")
    print("Columns:", list(df.columns))
    print("First 3 rows:")
    display(df.head(3))
```
